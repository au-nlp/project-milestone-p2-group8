### **EDA Recap**

#### **What We Did and Why:**

**1. Analyzed Our Target Variable (`overall_status`)**
*   **What:** We started by looking at the distribution of the `overall_status` column. We saw there were many categories ("Completed", "Recruiting", "Terminated", etc.) and the data was very imbalanced.
*   **Why:** Our project is to predict trial success, so we needed a clean, binary target variable. The raw statuses were too messy for this.
*   **Action:** We created a new `target` column. We defined **Success (1)** as "Completed" and **Failure (0)** as "Terminated," "Withdrawn," or "Suspended." This gives us a clear, but still imbalanced, target for our model.

**2. Selected Key Features for Analysis**
*   **What:** We chose a subset of structured features that we hypothesized would be good predictors: `phases`, `study_type`, `enrollment_count`, `lead_sponsor_class`, `sex`, `minimum_age`, and `maximum_age`.
*   **Why:** To focus our initial analysis on the most promising high-level features before diving into the more complex text data.

**3. Checked for Missing Values**
*   **What:** We calculated the percentage of missing data for our selected features.
*   **Why:** Most machine learning models can't handle missing values, so we needed to identify where the "holes" in our data are.
*   **Findings:** We found that `maximum_age` is missing in ~47% of rows, and `phases` is missing in ~21%. Other columns have very few missing values. This gives us a clear to-do list for our data cleaning pipeline.

**4. Analyzed Categorical Features**
*   **What:** We plotted the distributions of our categorical columns (`phases`, `study_type`, `lead_sponsor_class`, `sex`).
*   **Why:** To understand the composition of a "typical" trial in our dataset.
*   **Findings:**
    *   `phases`: We discovered that the `phases` column contains lists of values (e.g., `['PHASE2', 'PHASE3']`) and that many "missing" values are actually the string `"NA"` (likely meaning "Not Applicable" for observational studies).
    *   `lead_sponsor_class`: We found that the "OTHER" category is very large, which is a limitation we should be aware of.
    *   `study_type` & `sex`: These features are also imbalanced (e.g., more "Interventional" studies than "Observational"), but we concluded this is a reflection of reality and is still useful information for the model.

**5. Analyzed Numerical Features**
*   **What:** We investigated `enrollment_count`, `minimum_age`, and `maximum_age`. We first verified their data types, then generated summary statistics (`.describe()`), and finally plotted their distributions.
*   **Why:** To understand the scale, skewness, and data quality of our numerical predictors.
*   **Findings:**
    *   All three columns are heavily **right-skewed**, meaning most values are small, but a few extreme outliers exist (e.g., an `enrollment_count` in the millions). This was confirmed by comparing the `mean` to the `median`.
    *   The age columns contain clear **data entry errors** (e.g., a `minimum_age` of 730).
    *   A "zoomed-in" plot of `minimum_age` revealed it doesn't behave like a typical number. It's dominated by specific values like 18, 40, and 65, acting more like a **categorical feature**.
